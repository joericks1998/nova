```python
from nova_py import NERf, PARMesan
from transformers import BertTokenizer
import tensorflow as tf

tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
```

```python
text = "Transformers are amazing! DaVinci"
tokens = tokenizer.tokenize(text)
ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:", tokens)
print("Token IDs:", ids)
```

```python
ids_tensor = tf.constant([ids])
```

```python
pad_mask = tf.cast(ids_tensor != 0, tf.float32)
```

```python
type(ids_tensor)
```

```python
nerf_parameters = {
    "d_model": 1024,
    "num_transformers": 16,
    "num_features": 64,
    "vocabulary_size": tokenizer.vocab_size,
    "layerdrop": 0.2,
    "num_heads": 8,
    "dff": 4*1024,
    "top_p": 0.8,
    "dropout_rate": 0.2,
    "temperature": 1,
    "num_samples": 50
}
parmameters {
    "d_model": 1024,
    "num_transformers": 16,
    "input_size": 64,
    "output_size": tokenizer.vocab_size,
    "layerdrop": 0.2,
    "num_heads": 8,
    "dff": 4*1024,
    "top_p": 0.8,
    "dropout_rate": 0.2,
    "temperature": 1,
    "num_samples": 50,
    "nerf_params": nerf_params
}

    
```

```python
ner = NERf.Model(**parameters)
```

```python
tnsr= ner(ids_tensor, mask=pad_mask)
```
